```python
### what is if statement 

age=4 

if age>=20:
    print("over 18")
else:
    print("you are baby")
```

```python
### elif statement 
##allows you to check multiple conidtions 
```

```python
age=7 

if age<13:
    print("you are a child")
    
elif age<18:
    print("You are teenager")
else:
    print("ready to rock and roll")
```

```python
### nested conditional statements

## you can place one or more if else condition in elif condition 

## even odd or negative
```

```python
num=int(input("enter the number"))

if num >0:
    print("number is positibe")
    if num%2==0:
        print("Number is even")
    else:
        print("Number is odd")
elif num ==0:
    print("Its a zero")
else:
    print("Number is negative ")
    if num%2==0:
        print("Number is even")
    else:
        print("Number is odd")
```

```python
year=int(input("enter the year"))

if (year%4)==0:
    print("leap year")
else:
    ("non leap year")
```

```python
### simple calculator 
```

```python
num1= float(input("enter your first number"))
num2=float(input("enter your second number"))
operator=input("Enter the operation")

if operator=="+":
    result= num1+num2
    print("Result",result)
elif operator=="-":
    result=num1-num2
    print("Result",result)
elif operator=="*":
    result=num1*num2
    print("Result",result)
elif operator=="/":
    if num2!=0:
        result=num1/num2
        print("Result",result)
    else:
        print("cannot divide by zero")
else:
    print("Enter valid operation")

    
```
