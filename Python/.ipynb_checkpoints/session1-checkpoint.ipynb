# Keywords & Identifiers 

# Python Keywords

Python has **35 reserved keywords** (as of Python 3.11+).  
These keywords cannot be used as variable names, function names, or identifiers.

---

## List of Keywords

- `False`      - `await`      - `else`       - `import`     - `pass`  
- `None`       - `break`      - `except`     - `in`         - `raise`  
- `True`       - `class`      - `finally`    - `is`         - `return`  
- `and`        - `continue`   - `for`        - `lambda`     - `try`  
- `as`         - `def`        - `from`       - `nonlocal`   - `while`  
- `assert`     - `del`        - `global`     - `not`        - `with`  
- `async`      - `elif`       - `if`         - `or`         - `yield`  

---

## Key Points
- Keywords are **case-sensitive** (`True` is a keyword, but `true` is not).
- They are **reserved** and have predefined meaning in Python.
- Using them as identifiers will cause a **SyntaxError**.

👉 **Total Keywords:** 35


# Python Identifiers

**Identifiers** are the names given to variables, functions, classes, modules, or other objects in Python.  
They help to uniquely identify a name in the code.

---

## Rules for Identifiers

1. **Allowed Characters**  
   - Can contain **letters (a–z, A–Z)**, **digits (0–9)**, and **underscore (_)**.  
   - Example: `name`, `student_1`, `totalMarks`

2. **Must Not Start with a Digit**  
   - ❌ `123name` (invalid)  
   - ✅ `name123` (valid)

3. **Case-Sensitive**  
   - `Value`, `value`, and `VALUE` are **different identifiers**.

4. **Cannot Use Keywords**  
   - Python keywords like `for`, `class`, `if` **cannot** be used as identifiers.

5. **No Special Characters or Spaces**  
   - ❌ `student-name` or `total marks` (invalid)  
   - ✅ `student_name` or `totalMarks` (valid)

6. **Underscore Usage**  
   - `_name` : conventionally used for **private variables**.  
   - `__name__` : special identifiers used for **system-defined names** (e.g., `__init__`).  

---

## Examples

```python
# Valid identifiers
studentName = "John"
_age = 25
marks1 = 88
total_Marks = 450

# 6. Type conversion 

````python
a = input("Input first number: ")
b = input("Input second number: ")

try:
    # Try converting to float (works for int & float)
    a = float(a)
    b = float(b)

    # Now check types explicitly if you only want int/float
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        raise TypeError("Inputs must be int or float")

    print("Your answer is:", a + b)

except ValueError:
    print("❌ Invalid input! Please enter numbers only.")
except TypeError as e:
    print("❌", e)
````
