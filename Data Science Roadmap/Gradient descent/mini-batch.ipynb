```python
import random
class MBGDregressor:
    
    def __init__(self, learning_rate=0.01, epochs=100,batch_size=10):
        self.coef_=None
        self.intercept=None
        self.lr=learning_rate
        self.epochs=100
        self.batch_size=batch_size
    
    
    def fit(self,X_train,y_train):
        self.intercept= 0
        self.coef_= np.ones(X_train.shape[1])
        
        for i in range(self.epochs):
            #update all the coeff and intercept
            for j in range(int(X_train.shape[0]/self.batch_size)):
                
                idx=random.sample(range(X_train.shape[0]),self.batch_size)
                
                y_hat=np.dot(X_train[idx],self.coef_) + self.intercept
                
                intercept_der= -2 * np.mean(y_train[idx]-y_hat)
                
                self.intercept=self.intercept- (self.lr * intercept_der)
            
            
            
                coef_der= -2 * np.dot((y_train[idx]-y_hat),X_train[idx])
                self.coef_=self.coef_- (self.lr * coef_der)
        
        print(self.coef_,self.intercept)
        
    
    def predict(self,X_test):
        return np.dot(X_test,self.coef_)+ self.intercept
        
        
```

```python
reg=MBGDregressor(0.01,100,int(X_train.shape[0]/10))
```

```python
reg.fit(X_train,y_train)
```

```python
y_pred= reg.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```
